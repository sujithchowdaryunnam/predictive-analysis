```python
import pandas as pd

# Load datasets
matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

def show_options():
    print("""
📊 IPL Analysis Options:
1. Team performance (wins/losses)
2. Player batting stats
3. Player bowling stats
4. Head-to-head between two teams
5. Match details by season or venue
""")

def get_close_match(name, all_names):
    # Case-insensitive matching with partial match support
    matches = [n for n in all_names if name.lower() in n.lower()]
    return matches

def team_performance():
    team = input("Enter team name: ").strip()
    teams = set(matches['team1']).union(set(matches['team2']))
    close = get_close_match(team, teams)
    if not close:
        print("❌ Team not found. Try again.")
        return
    team = close[0]  # Pick best match
    total_matches = matches[(matches['team1'] == team) | (matches['team2'] == team)]
    wins = matches[matches['winner'] == team]
    print(f"\n📌 {team} played {len(total_matches)} matches and won {len(wins)}.")
    if not wins.empty:
        print(f"🏆 Win percentage: {len(wins) / len(total_matches) * 100:.2f}%")

def batting_stats():
    player = input("Enter batsman name: ").strip()
    batsmen = deliveries['batsman'].unique()
    matches_found = get_close_match(player, batsmen)
    if not matches_found:
        print("❌ No matching batsman found. Check spelling.")
        return
    player = matches_found[0]
    df = deliveries[deliveries['batsman'] == player]
    total_runs = df['batsman_runs'].sum()
    balls = df.shape[0]
    sr = (total_runs / balls * 100) if balls > 0 else 0
    print(f"\n🏏 {player} - Runs: {total_runs}, Balls: {balls}, Strike Rate: {sr:.2f}")

def bowling_stats():
    player = input("Enter bowler name: ").strip()
    bowlers = deliveries['bowler'].unique()
    matches_found = get_close_match(player, bowlers)
    if not matches_found:
        print("❌ No matching bowler found. Check spelling.")
        return
    player = matches_found[0]
    df = deliveries[deliveries['bowler'] == player]
    wickets = df[df['dismissal_kind'].notnull() & (df['dismissal_kind'] != 'run out')].shape[0]
    runs = df['total_runs'].sum()
    overs = df.shape[0] / 6
    econ = (runs / overs) if overs > 0 else 0
    print(f"\n🎯 {player} - Wickets: {wickets}, Economy: {econ:.2f}, Overs: {overs:.1f}")

def head_to_head():
    team1 = input("Enter Team 1: ").strip()
    team2 = input("Enter Team 2: ").strip()
    teams = set(matches['team1']).union(set(matches['team2']))
    t1_matches = get_close_match(team1, teams)
    t2_matches = get_close_match(team2, teams)
    if not t1_matches or not t2_matches:
        print("❌ One or both teams not found.")
        return
    team1, team2 = t1_matches[0], t2_matches[0]
    h2h = matches[((matches['team1'] == team1) & (matches['team2'] == team2)) |
                  ((matches['team1'] == team2) & (matches['team2'] == team1))]
    print(f"\n📋 Total Matches between {team1} and {team2}: {len(h2h)}")
    print(h2h['winner'].value_counts())

def match_details():
    season = input("Enter season/year (or press Enter to skip): ").strip()
    venue = input("Enter venue (or press Enter to skip): ").strip()
    df = matches.copy()
    if season:
        df = df[df['season'] == int(season)]
    if venue:
        df = df[df['venue'].str.contains(venue, case=False)]
    print(f"\n📅 Matches Found: {len(df)}")
    print(df[['date', 'team1', 'team2', 'winner', 'venue']].head(10))

# ---- Main Program ----
while True:
    show_options()
    choice = input("Enter choice (1-5, q to quit): ").strip().lower()

    if choice == '1':
        team_performance()
    elif choice == '2':
        batting_stats()
    elif choice == '3':
        bowling_stats()
    elif choice == '4':
        head_to_head()
    elif choice == '5':
        match_details()
    elif choice == 'q':
        print("👋 Exiting. Goodbye!")
        break
    else:
        print("❌ Invalid choice. Try again.")
```
