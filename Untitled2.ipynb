```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")

# Load your dataset
file_path = 'stock_data.csv'
data = pd.read_csv(file_path)

# Convert date to datetime
data['Date'] = pd.to_datetime(data['Date'])

# Ask for user input
stock_name = input("Enter the stock name to predict (as in dataset): ").upper()
future_days = int(input("Enter number of days to predict into the future: "))

# Filter for the selected stock
if 'Stock' in data.columns:
    stock_data = data[data['Stock'].str.upper() == stock_name].copy()
else:
    raise ValueError("❌ 'Stock' column not found in the dataset.")

# Sort by date
stock_data.sort_values('Date', inplace=True)

# Reset index
stock_data.reset_index(drop=True, inplace=True)

# Use only the 'Close' column for prediction
stock_data = stock_data[['Date', 'Close']].copy()
stock_data['Prediction'] = stock_data[['Close']].shift(-future_days)

# Prepare the features (X) and labels (y)
X = np.array(stock_data.drop(['Prediction', 'Date'], axis=1))[:-future_days]
y = np.array(stock_data['Prediction'])[:-future_days]

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict future prices
X_future = np.array(stock_data.drop(['Prediction', 'Date'], axis=1))[-future_days:]
predictions = model.predict(X_future)

# Prepare dates for visualization
future_dates = pd.date_range(start=stock_data['Date'].iloc[-future_days], periods=future_days).to_list()

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(stock_data['Date'], stock_data['Close'], label='Historical Close Price')
plt.plot(future_dates, predictions, label='Predicted Price', color='red')
plt.title(f"{stock_name} Stock Price Prediction ({future_days} Days)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Load and clean the dataset
file_path = 'stock_data.csv'
data = pd.read_csv(file_path)

# Rename index column to 'date' (could be Unnamed: 0 or similar)
data.rename(columns={data.columns[0]: 'date'}, inplace=True)

# Try to convert to datetime; fallback to integer index if it fails
try:
    data['date'] = pd.to_datetime(data['date'])
except:
    data['date'] = pd.RangeIndex(start=0, stop=len(data), step=1)

# Detect stock columns (everything except 'date')
stock_columns = [col for col in data.columns if col.lower() != 'date']
stock_map = {col.lower(): col for col in stock_columns}  # mapping lowercase -> actual

# Show available stock options
print("📈 Available stocks:", stock_columns)

# User input
user_input = input("Enter the stock column to predict (e.g., stock_1): ").strip().lower()
future_days = int(input("Enter number of days to predict into the future: "))

# Validate input
if user_input not in stock_map:
    raise ValueError(f"❌ '{user_input}' not found in dataset columns.")

# Get actual column name
stock_name = stock_map[user_input]

# Prepare data
df = data[['date', stock_name]].copy()
df.rename(columns={stock_name: 'close'}, inplace=True)
df['prediction'] = df['close'].shift(-future_days)

# Features and labels
X = np.array(df[['close']])[:-future_days]
y = np.array(df['prediction'])[:-future_days]

# Split into training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict future values
X_future = np.array(df[['close']])[-future_days:]
predictions = model.predict(X_future)

# Create future dates
future_dates = pd.date_range(start=df['date'].iloc[-future_days], periods=future_days).to_list()

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['close'], label='Historical Price')
plt.plot(future_dates, predictions, label='Predicted Price', color='red')
plt.title(f"{stock_name} Price Prediction ({future_days} Days)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
